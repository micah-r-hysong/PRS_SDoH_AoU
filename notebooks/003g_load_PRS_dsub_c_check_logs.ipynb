```python
%%bash

#!/bin/bash

# Setup
BUCKET_BASE="gs://fc-secure-784622b3-a3b0-42a1-a555-372ab8841bd6"
LOG_ROOT="${BUCKET_BASE}/dsub/logs"
LOCAL_LOG_DIR="./local_logs"
#mkdir -p "$LOCAL_LOG_DIR"

# Clean up old logs if any
#rm -f "$LOCAL_LOG_DIR"/*

# Get all .log files
#log_paths=$(gsutil ls -r "${LOG_ROOT}/**/*task-None.log")

# Download all logs
#for log_path in $log_paths; do
  #gsutil cp "$log_path" "$LOCAL_LOG_DIR/"
#done

#!/bin/bash

LOG_DIR="./local_logs"
OUTPUT="predictors_matrix.csv"
declare -A counts
declare -A chr_set
declare -A disease_set

for log in "$LOG_DIR"/*.log; do
  # Extract logging line: "Logging to PRS_chr6_atrial_fibrilation.log"
  log_line=$(grep -m1 "Logging to PRS_" "$log")
  echo $log_line
  
  # Extract chromosome and disease
  if [[ "$log_line" =~ PRS_(chr[0-9XY]+)_(.+)\.log ]]; then
    chr="${BASH_REMATCH[1]}"
    disease="${BASH_REMATCH[2]}"
    
    chr_set["$chr"]=1
    disease_set["$disease"]=1

    # Extract valid predictor count
    predictors=$(grep -oP "[0-9]+(?= valid predictor\(s\) loaded)" "$log" | head -n 1)
    predictors=${predictors:-0}

    key="${chr}|${disease}"
    counts["$key"]=$predictors
  fi
done

# Create header row
{
  printf "chromosome"
  for disease in $(printf "%s\n" "${!disease_set[@]}" | sort); do
    printf ",%s" "$disease"
  done
  echo

  for chr in $(printf "%s\n" "${!chr_set[@]}" | sort -V); do
    printf "%s" "$chr"
    for disease in $(printf "%s\n" "${!disease_set[@]}" | sort); do
      key="${chr}|${disease}"
      printf ",%s" "${counts[$key]:-0}"
    done
    echo
  done
} > "$OUTPUT"

echo "✅ Predictor matrix saved to $OUTPUT"
```

```python
%%bash

#!/bin/bash

# Define an array of disease names
#diseases=("asthma" "atrial_fibrilation" "breast_cancer" "ckd" "coronary_heart_disease" "hypercholesterolemia"
          #"prostate_cancer" "t1d" "t2d")
          
diseases=("asthma" "atrial_fibrilation")

# Base GCS log directory
log_directory="gs://fc-secure-784622b3-a3b0-42a1-a555-372ab8841bd6/dsub/logs/*/micah-hysong/*task-None.log"

# Local folder for logs
local_log_directory="./local_logs"
mkdir -p "$local_log_directory"

# Get all logs from GCS
log_files=$(gsutil ls "$log_directory")

# Copy all logs locally
for file in $log_files; do
  gsutil -m cp "$file" "$local_log_directory/"
done

# Loop over each disease
for disease in "${diseases[@]}"; do
  echo "Processing disease: $disease"
  weights="_${disease}"  # Expected pattern in log file

  processed_chromosomes=()

  for log_file in "$local_log_directory"/*; do
    if grep -q "Delocalizing OUTPUT_PATH" "$log_file" && grep -q "$weights" "$log_file"; then
      
      # Extract chromosome info based on disease name
      chromosome=$(grep -o "chr[0-9XY]\+${weights}" "$log_file" | grep -o "chr[0-9XY]\+" | sort -u)

      if [[ -z "$chromosome" ]]; then
        echo "Chromosome not found in $log_file"
        continue
      fi

      if [[ " ${processed_chromosomes[@]} " =~ " ${chromosome} " ]]; then
        continue
      fi

      processed_chromosomes+=("$chromosome")
      echo "Chromosome for $log_file: $chromosome"

      # Extract predictor count
      predictor_count_list=($(grep -oP "(?<= )[0-9]+(?= valid predictors loaded)" "$log_file" || echo 0))

      echo "Final Predictor Count List for ${chromosome}_${disease}: ${predictor_count_list[@]}"

      # Save to CSV
      csv_file="./${chromosome}_${disease}_variants_incorporated.csv"
      printf "%s\n" "${predictor_count_list[@]}" > "$csv_file"

      # Upload CSV to GCS
      gsutil cp "$csv_file" "gs://fc-secure-784622b3-a3b0-42a1-a555-372ab8841bd6/data/"
    fi
  done
done


```
